```python
window1=tkinter.Tk()
window1.title('Window1')

var1=tkinter.IntVar()
tkinter.Checkbutton(window1, text='Male', variable=var1).grid(row=0, sticky='W')

var2=tkinter.IntVar()
tkinter.Checkbutton(window1, text='Female', variable=var2).grid(row=1, sticky='W')

window1.mainloop()
```

```python
window2=tkinter.Tk()
window2.title('Window2')

tkinter.Label(window2,text='First Name').grid(row=0)
tkinter.Label(window2,text='Lat Name').grid(row=1)

entry1=tkinter.Entry(window2)
entry2=tkinter.Entry(window2)

entry1.grid(row=0,column=1)
entry2.grid(row=1,column=1)

window2.mainloop()
```

```python
win=tkinter.Tk()
msg='Welcome to GUI Programming'
msgVar=tkinter.Message(win,text=msg)
msgVar.config(bg='blue',justify='center')
msgVar.pack()
win.mainloop()
```

```python
dow=tkinter.Tk()
var=tkinter.IntVar()
tkinter.Radiobutton(dow,text='True',variable=var,value=1).pack(anchor='w')
tkinter.Radiobutton(dow,text='False',variable=var,value=2).pack(anchor='w')
dow.mainloop()
```
