### Map Marker

```python
#Using Map Marker
map1 = folium.Map(location=[14.655272431277755, 121.06720290942367],
                  zoom_start=15, #setting the initial zoom of the map
                  tiles="OpenStreetMap")

#creating a marker
folium.Marker(
    [14.656685972434557, 121.06959063008526], #exact marker location
    popup="<i>Melchor Hall</i>", #popup title of the marker this uses html encoding for the text formatting but can also be left just as is
    tooltip="I'm a marker!" #display when hovering at the marker
).add_to(map1) #adding the marker to the map

#another marker
folium.Marker(
    [14.655034710248392, 121.07281801330198],
    popup="UP Grandstand",
    tooltip="Click here",
    icon=folium.Icon(color="pink", icon="info-sign") #set the icon type and color of the marker
).add_to(map1)

#another marker
folium.Marker(
    [14.655160983342945, 121.07107515124595],
    popup="<b>UP Main Library</b>",
    tooltip="Click here",
    icon=folium.Icon(color='pink', icon="book", prefix='fa')
    #you can search icons in font awesome (https://fontawesome.com/v4/icons/) but need to add an argument, 'prefix'
).add_to(map1)

map1
```

### Circle Markers

```python
#Using Circle Markers
map2 = folium.Map(location=[14.655272431277755, 121.06720290942367],
                  zoom_start=15, #setting the initial zoom of the map
                  tiles="OpenStreetMap")

#creating a circle
folium.Circle(
    radius=100, #setting circle radius
    location=[14.656685972434557, 121.06959063008526],
    popup="Melchor Hall",
    color="crimson",
    fill=False, #setting if circle is filled or not
).add_to(map2)

#another type of circle marker -- difference with 'circle' alone is that circlemarker resizes with zooming in or out
folium.CircleMarker(
    location=[14.655034710248392, 121.07281801330198],
    radius=50,
    popup="Grandstand",
    tooltip="This is a place in UP",
    color="#3186cc",
    fill=True,
    fill_color="#3186cc"
).add_to(map2)

map2
```

### Polygon Markers

```python
#Polygon Markers
folium.RegularPolygonMarker(location=[14.655160983342945, 121.07107515124595],
                            popup = 'Main Library',
                            fill_color = '#00ff40',
                            number_of_sides = 8,
                            radius = 100).add_to(map2)

map2
```

### Pop Up Information & Moving Markers

```python
map1.add_child(folium.LatLngPopup()) #add a popup of location (lat/long) every time you click at the map

map1.add_child(folium.ClickForMarker(popup="A new POI")) #adding a feature where you can add a marker to the map!

folium.Marker(location=[14.655033849295586, 121.0720395415908], draggable=True).add_to(map1) #make a movable marker

map1
```

```python
#sample using GeoJson available online from: https://github.com/jehyunlee

m = folium.Map(location=[37.566345, 126.977893], zoom_start=8)

jsonData = f'https://raw.githubusercontent.com/jehyunlee/data-snippet/master/Korea_gu_20.json'        # Reading Admin. Districts

style_function = {
                 'fillColor': 'white',
                 'color':'black'              # edge color change : black
                }
folium.GeoJson(jsonData, name='json_data',
               style_function=lambda x: style_function        # style_function has to be a function which calls dictionary
              ).add_to(m)        # Overlay on map

m
```

## GeoPandas

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import geopandas as gpd
import numpy as np
```

```python
brgy_bounds = gpd.read_file('qc_brgys.gpkg')
brgy_bounds.head()
```

```python
brgy_bounds.set_index('Bgy_Name', drop=False, inplace=True)
brgy_bounds.head()
```

```python
#generate a random integer between 0-100 and add as 'value' in brgy_bounds.
brgy_bounds['values'] = brgy_bounds.apply(lambda _: np.random.randint(0,100), axis=1)
brgy_bounds.head()
```

```python
def get_random_int():

  return np.random.randint(0,100)

brgy_bounds['values'] = brgy_bounds.apply(lambda _: get_random_int(), axis=1)
```

```python
brgy_bounds.head()
```

```python
def get_diff(x):
  return x-10

brgy_bounds['diff'] = brgy_bounds['values'].apply(lambda element: get_diff(element))
```

```python
brgy_bounds.head()
```

```python
brgy_bounds.plot(column='values',
                 legend=True,
                 cmap="mako",
                 vmin=0, vmax=100,
                 legend_kwds={"label": "Random Values", "orientation": "horizontal"})
plt.savefig('sample_map1.png', format='png', bbox_inches='tight', dpi=100)
plt.clf()
plt.close()
```
